```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

def crawl_jobkorea(search_keyword):
    base_url = "https://www.jobkorea.co.kr"
    search_url = f"{base_url}/Search/?stext={search_keyword}"
    
    headers = {
        'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4472.124 Safari/537.36'
    }
    
    try:
        response = requests.get(search_url, headers=headers)
        response.raise_for_status()
        soup = BeautifulSoup(response.text, 'html.parser')
        
        # 채용공고 리스트 추출
        recruit_list = soup.select('article.list-item')
        
        data = {
            'site': [],
            'Col_Company': [],
            'Col_Recruit': [],
            'Col_detail': [],
            'Col_url': []
        }
        
        for recruit in recruit_list:
            # 회사명
            company_elem = recruit.select_one('a.corp-name-link.dev-view')
            company = company_elem.get_text(strip=True) if company_elem else 'N/A'
            
            # 채용공고 제목
            title_elem = recruit.select_one('a.information-title-link.dev-view')
            title = title_elem.get_text(strip=True) if title_elem else 'N/A'
            
            # 상세 조건 (경력, 학력 등)
            detail_elem = recruit.select_one('ul.chip-information-group')
            detail = detail_elem.get_text(strip=True) if detail_elem else 'N/A'
            
            # 채용공고 URL
            relative_url = title_elem['href'] if title_elem and 'href' in title_elem.attrs else ''
            full_url = base_url + relative_url if relative_url.startswith('/') else relative_url
            
            # 데이터 저장
            data['site'].append('Job_Korea')
            data['Col_Company'].append(company)
            data['Col_Recruit'].append(title)
            data['Col_detail'].append(detail)
            data['Col_url'].append(full_url)
        
        # 데이터프레임 생성
        df = pd.DataFrame(data)
        df.to_csv('data_jobkorea.csv', index=False, encoding='utf-8-sig')
        return df
    
    except Exception as e:
        print(f"Error occurred: {e}")
        return pd.DataFrame()

# '데이터분석'으로 검색한 채용공고 크롤링
df = crawl_jobkorea('데이터분석')
# 결과 출력
display(df)
```

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

def crawl_jobkorea(search_keyword):
    base_url = "https://www.jobkorea.co.kr"
    search_url = f"{base_url}/Search/?stext={search_keyword}"
    
    headers = {
        'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4472.124 Safari/537.36'
    }
    
    try:
        response = requests.get(search_url, headers=headers)
        response.raise_for_status()
        soup = BeautifulSoup(response.text, 'html.parser')
        
        # 채용공고 리스트 추출
        recruit_list = soup.select('article.list-item')
        
        data = {
            'site': [],
            'Col_Company': [],
            'Col_Recruit': [],
            'Col_detail': [],
            'Col_url': []
        }
        
        for recruit in recruit_list:
            # 회사명
            company_elem = recruit.select_one('div.list-section-corp a.corp-name-link')
            company = company_elem.get_text(strip=True) if company_elem else 'N/A'
            
            # 채용공고 제목
            title_elem = recruit.select_one('div.information-title a.information-title-link')
            title = title_elem.get_text(strip=True) if title_elem else 'N/A'
            
            # 상세 조건 (경력, 학력, 지역 등)
            detail_elems = recruit.select('ul.chip-information-group li.chip-information-item')
            details = [elem.get_text(strip=True) for elem in detail_elems] if detail_elems else []
            detail = ' | '.join(details) if details else 'N/A'
            
            # 채용공고 URL
            title_link = recruit.select_one('div.information-title a.information-title-link')
            relative_url = title_link['href'] if title_link and 'href' in title_link.attrs else ''
            full_url = base_url + relative_url if relative_url.startswith('/') else relative_url
            
            # 데이터 저장
            data['site'].append('Job_Korea')
            data['Col_Company'].append(company)
            data['Col_Recruit'].append(title)
            data['Col_detail'].append(detail)
            data['Col_url'].append(full_url)
        
        # 데이터프레임 생성
        df = pd.DataFrame(data)
        df.to_csv('data_jobkorea.csv', index=False, encoding='utf-8-sig')
        return df
    
    except Exception as e:
        print(f"Error occurred: {e}")
        return pd.DataFrame()

# '데이터분석'으로 검색한 채용공고 크롤링
df = crawl_jobkorea('데이터분석')
# 결과 출력
display(df)
```
