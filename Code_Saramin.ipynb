```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

def crawl_saramin(search_keyword):
    base_url = "https://www.saramin.co.kr/"
    search_url = f"{base_url}/zf_user/search?search_area=main&search_done=y&search_optional_item=n&searchType=recently&searchword={search_keyword}"
    
    headers = {
        'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4472.124 Safari/537.36'
    }
    
    try:
        response = requests.get(search_url, headers=headers)
        response.raise_for_status()
        soup = BeautifulSoup(response.text, 'html.parser')
        
        # 채용공고 리스트 추출
        recruit_list = soup.select('div.item_recruit')
        
        data = {
            'site': [],
            'Col_Company': [],
            'Col_Recruit': [],
            'Col_detail': [],
            'Col_url': []
        }
        
        for recruit in recruit_list:
            # 회사명
            company_elem = recruit.select_one('strong.corp_name a')
            company = company_elem.get_text(strip=True) if company_elem else 'N/A'
            
            # 채용공고 제목
            title_elem = recruit.select_one('h2.job_tit a')
            title = title_elem.get_text(strip=True) if title_elem else 'N/A'
            
            # 상세 조건 (지역, 경력, 학력, 고용형태)
            condition_elems = recruit.select('div.job_condition span')
            details = []
            for elem in condition_elems:
                # Remove nested <a> tags if they exist
                a_tag = elem.select_one('a')
                if a_tag:
                    details.append(a_tag.get_text(strip=True))
                else:
                    details.append(elem.get_text(strip=True))
            detail = ' '.join(details) if details else 'N/A'
            
            # 채용공고 URL
            title_link = recruit.select_one('h2.job_tit a')
            relative_url = title_link['href'] if title_link and 'href' in title_link.attrs else ''
            full_url = base_url + relative_url if relative_url.startswith('/zf_user') else relative_url
            
            # 데이터 저장
            data['site'].append('Saramin')
            data['Col_Company'].append(company)
            data['Col_Recruit'].append(title)
            data['Col_detail'].append(detail)
            data['Col_url'].append(full_url)
        
        # 데이터프레임 생성
        df = pd.DataFrame(data)
        df.to_csv('data_saramin.csv', index=False, encoding='utf-8-sig')
        return df
    
    except Exception as e:
        print(f"Error occurred: {e}")
        return pd.DataFrame()

# '데이터분석'으로 검색한 채용공고 크롤링
df = crawl_saramin('데이터분석')
# 결과 출력
display(df)
```
